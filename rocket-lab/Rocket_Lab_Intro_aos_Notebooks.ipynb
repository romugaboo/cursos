# 3 Dicas para submissão
- Assegure-se de salvar seu notebook antes de submeter, é muito comum que ao executar a célula de submissão o nosso corretor não “enxergar” as últimas alterações do seu código.
- Reinicie o ambiente de execução (tanto jupyter como colab), pois se você estiver utilizando-o a muito tempo as variáveis podem estar guardando valores de execuções passadas que não correspondem ao seu código atual. Ao reiniciar você irá apagar os valores de todas as variáveis chave, permitindo que a correção passe pelo seu código “do zero” e que elas recebam os valores adequados para a submissão.





Como boa prática para submissão sempre reinicie o ambiente de execução, salve o notebook e apenas após isso, execute o comando de submissão.

PS: Em alguns casos, se você estiver utilizando o colab, não é apenas a memória do seu notebook que está ‘confusa’ mas também a do seu navegador. Caso você não saiba, ele também tem uma memória para se “lembrar” das páginas e links que você visitou. Em caso de erro também pode ser interessante limpar a memória do seu navegador – cada um (*Chrome, Firefox, Edge...*) tem sua maneira própria para isso, recomendamos que busque a documentação para aquele de sua preferência.
