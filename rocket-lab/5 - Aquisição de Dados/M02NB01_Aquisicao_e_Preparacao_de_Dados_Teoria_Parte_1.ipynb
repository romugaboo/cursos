## 1. Introdução

Olá, pessoal! Nessa aula nos aporfundaremos ao Data Science Pipeline! Nesta lição estamos falando de dois tópicos, **Aquisição de Dados** e **Limpeza de Dados**, os passos 4 e 5 de nosso Pipeline de Ciência de Dados.

Aquisição de Dados é o ato de coletar dados para que possamos aplicar modelos estatísticos a ele. Temos 4 métodos principais para fazer isso:
1. Consulta (*querying*), obtenção de dados de bancos de dados. O mais comum se o seu projeto estiver em uma organização.
2. Obter dados de fontes externas gratuitas, como Kaggle e Google Datasets.
3. Raspagem (*scraping*) a partir da web, ou seja, usamos Python para acessar websites e recuperar informações a partir deles usando [rastreadores (*crawlers*)](https://en.wikipedia.org/wiki/Web_crawler) e [métodos de raspagem](https://en.wikipedia.org/wiki/Web_scraping).

Depois, quando tivermos os dados, devemos limpá-los. Os dados, mais frequente que raro, são confusos, ausentes e não muito bem documentados. Portanto, temos que diagnosticar nossos dados e determinar se e como podemos limpá-los.

### 2.2. *Flat files*

*Flat files* são aqueles que têm apenas conteúdo de texto estruturado no sentido de uma única tabela com um registro por linha com colunas delimitadas por um carácter fixo. As extensões mais comuns dos *flat files* são .txt e .csv. TXT é uma abreviação para texto, enquanto CSV significa Comma Separated Values (Valores Separados por Vírgula).

**Baixe o arquivo zip mencionado acima** e descompacte-o na pasta ***data***. Ele tem alguns arquivos que vamos usar ao longo desta lição.

Agora podemos importar o Pandas e ler um arquivo simples.

````python
import pandas as pd

pokemon_df = pd.read_csv('dados/pokemon.csv')
pokemon_df.head()
````

Bem fácil, não? Basta utilizar o método [```pd.read_csv()```](https://pandas.pydata.org/docs/reference/api/pandas.read_csv.html) passando o caminho do arquivo como o primeiro argumento. Mas isso não é novidade para você. Vamos tentar isso agora com o arquivo winequality-red.csv.

````python
wine_red_df = pd.read_csv('dados/winequality-red.csv')
wine_red_df.head()
````

Parece que algo está errado. Abra o arquivo em um editor de texto e tente entender o que aconteceu. Você pode abrir o pokemon.csv para tentar perceber a diferença também.

Você conseguiu perceber o que aconteceu? Este último arquivo é separado por ponto-e-vírgula, embora o formato seja **valores separados por vírgula**. Pode parecer um pouco estranho, mas os arquivos txt e csv podem ter qualquer caractere como separador. Para lê-lo, é necessário especificar esse caractere utilizando o parâmetro ```sep``` ou ```delimiter```.

````python
wine_red_df = pd.read_csv('dados/winequality-red.csv', sep=';')
wine_red_df.info()
wine_red_df.head()
````

Se olharmos com atenção, ainda há algo errado aqui. Há algumas colunas que parecem ter apenas números *float*, mas não foram listadas como *float* quando pedimos por *info*. Isso porque o caractere decimal padrão de ponto flutuante é um ponto, mas aqui temos vírgulas. Devemos mudar isso também utilizando o parâmetro ```decimal```.

````python
wine_red_df = pd.read_csv('dados/winequality-red.csv', delimiter=';', decimal=',')
wine_red_df.info()
````

Assim é melhor, mas a acidez volátil, densidade e álcool ainda estão sendo lidos como objeto. É porque existem alguns valores acima de 1000 que são escritos com um ponto que separa os milhares. Como você acha que podemos resolver isso?

````python
wine_red_df = pd.read_csv('dados/winequality-red.csv', delimiter=';', decimal=',', thousands='.')
wine_red_df.info()
````

Muito simples, não é?

Como você pode ver, o ```pd.read_csv``` tem uma tonelada de parâmetros que podem ajudar na leitura dos arquivos. Abaixo temos uma lista dos parâmetros que vimos e alguns outros parâmetros importantes, essa lista não é exaustiva, mas você pode ver a lista completa na [documentação](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.read_csv.html).

- **delimiter**/**sep** - especifica o caractere que está sendo usado como separador de colunas no arquivo que está sendo lido;
- **decimal** - especifica o caractere que está sendo usado como separador de ponto flutuante no arquivo que está sendo lido;
- **thousands** - especifica o caractere sendo usado como separador de milhares no arquivo que está sendo lido;
- **names** - lista de nomes de colunas a serem usadas;
- **usecols** - lista de subconjuntos de colunas a serem mantidas no conjunto de dados ou uma função a ser avaliada em cada coluna;
- **skiprows** - números de linhas a pular (indexados a 0) ou número de linhas a pular (int) no início do arquivo;
- **nrows** - número de linhas a ler, útil para a leitura de peças de arquivos grandes;

Você pode vê-los em ação abaixo.

````python
wine_red_df = pd.read_csv('dados/winequality-red.csv', delimiter=';', decimal=',', thousands='.',
                      names=['f_acidity', 'v_acidity', 'c_acid', 'res_sugar', 'chlorides', 'free_sulf_dioxide',
                             'tot_sulfur_dioxide', 'density', 'pH', 'sulphates', 'alcohol', 'quality'],
                     usecols=['f_acidity', 'v_acidity', 'density', 'pH', 'quality'],
                     skiprows=3,
                     nrows=5)
wine_red_df.info()
````

### 2.3. Arquivos Excel

Usamos arquivos Excel a todo momento. Cada arquivo tem a extensão .xlsx, é chamado de Workbook e pode conter várias Planilhas (*Spreadsheets*).

Para abrir uma Pasta de Trabalho com pandas, utilizamos o método [```ExcelFile```](https://pandas.pydata.org/docs/reference/api/pandas.ExcelFile.parse.html). Isso retorna um objeto do qual podemos ver todas as planilhas disponíveis com o campo ```sheet_names``` e obter qualquer um deles utilizando o método ```parse```. Vamos ver isso funcionando.

````python
# Abre o Workbook e mostra o nome das planilhas
file = 'dados/winequality.xlsx'
wine_data = pd.ExcelFile(file)
print(wine_data.sheet_names)

# Interpreta (parse) as planilhas por nome ou index
wine_white_df = wine_data.parse('white')
wine_red_df = wine_data.parse(1)

wine_white_df.info()
wine_red_df.info()
````

Entende como funciona? É bom mencionar que você pode analisar uma planilha de cálculo por nome ou índice.

Além disso, assim como  ```read_csv```, você pode utilizar alguns parâmetros com ```parse```, demonstrado abaixo. A lista completa de parâmetros está na [documentação](https://pandas.pydata.org/docs/reference/api/pandas.ExcelFile.parse.html).

````python
wine_white_df = wine_data.parse(0,
                                usecols=[0, 1, 7, 8, 11],
                                skiprows=[0],
                                names=['f_acidity', 'v_acidity', 'density', 'pH', 'quality'])

wine_white_df.info()
````

Outra maneira de ler arquivos excel é com o método ```read_excel```. Você pode simplesmente ler o arquivo e o parâmetro ```sheet_name```.

````python
another_wine_df = pd.read_excel(file, sheet_name='white')
another_wine_df.info()
````

### 2.5. Pickle

Muito bem, aprendemos um pouco sobre a leitura de dados em um formato de tabela. Mas e se quisermos armazenar e ler um dicionário? Só para lembrar, um dicionário é uma coleção de pares de valores-chave, onde o valor pode até ser outro dicionário, dificultando o armazenamento como uma tabela. Tomemos um exemplo abaixo, onde criamos um dicionário ```people``` que têm John e Marie.

````python
people = {1: {'name': 'John',  'age': '27', 'sex': 'Male', 'siblings':['Johnson', 'Johan']},
          2: {'name': 'Marie', 'age': '22', 'sex': 'Female', 'dayofbirth': 12}}

print(people)
````

Seria difícil armazená-lo em uma tabela, já que John e Marie têm chaves diferentes, especialmente com a chave  ```siblings``` tendo uma lista como um valor. Aqui entra **pickle**.

Pickle permite que você escreva cada objeto Python em um arquivo e então consiga lê-los de volta quando precisar.

Para escrever, precisamos abrir um arquivo (python cria o arquivo se ele não existir) e utilizar ```dump```. É importante, toda vez que abrirmos um arquivo, fechá-lo após sua utilização.

Então para ler, abra o arquivo existente, utilize ```load``` e feche o arquivo novamente.

Para abrir um arquivo, usamos a função ```open```, passando o nome do arquivo e o modo. Por enquanto, você precisa saber os modos 'r' para ler, 'w' para escrever e 'a' para anexar. Além disso, o modo 'b' é usado junto com esses em pickle para salvar o arquivo como binário.

Veja o exemplo abaixo para entender melhor.

````python
# importando pickle
import pickle

# abrindo o arquivo 'pickle_file.pkl' em modo 'w' para escrita e 'b' para binário
pickle_file_write = open('dados/pickle_file.pkl', 'wb')

# dumping, isto é, salvando o objeto no arquivo
pickle.dump(people, pickle_file_write)

# fechando o arquivo
pickle_file_write.close()


# abrindo o arquivo 'pickle_file.pkl' em modo 'r' para leitura e 'b' para binário
pickle_file_read = open('dados/pickle_file.pkl', 'rb')

# carregando o objeto
loaded_people = pickle.load(pickle_file_read)

# fechando o arquivo
pickle_file_read.close()

# compara ambos objetos
print('People write:', people)
print('People load: ', loaded_people)

# ve o arquivo .pkl criado
print('\nFiles:', os.listdir(wd_data))
````

Uma última coisa sobre pickle e arquivos em geral.

É realmente importante que você sempre feche seus arquivos, para que outras pessoas ou softwares possam acessá-los. Para ajudá-lo com isso, o Python tem a declaração ```with``` , que pode ser usada para carregar um arquivo, utilizá-lo dentro do bloco ```with``` e fechar automaticamente o arquivo, assim:

```python
with expression [as variable]:
    statements
```

Vamos ver um exemplo.

````python
# salvando o objeto dentro do with
with open('dados/pickle_file_with.pkl', 'wb') as file_write:
    pickle.dump(people, file_write)

# carregando o objeto dentro do with
with open('dados/pickle_file_with.pkl', 'rb') as file_read:
    loaded_people_with = pickle.load(file_read)

# compara ambos objetos
print('People write:     ', people)
print('People load with: ', loaded_people_with)

# ver o arquivo .pkl criado
print('\nFiles:', os.listdir(wd_data))

# tenta usar o arquivo fora do bloco with para ter certeza de que já está fechado (deveria nos dar um erro)
loaded_people_with = pickle.load(file_read)
````

Vê como recebemos um erro que nos mostra que o arquivo já está fechado? Isso é muito útil!

Você deve sempre usar ```with``` enquanto trabalha com arquivos!
