#### Exercício 2.9

Faça um ```LEFT JOIN``` para trazer todos os albuns de cada artista

**Dica**: Os apelidos no SQL são usados para dar um nome temporário a uma tabela ou a uma coluna em uma tabela. Os apelidos são frequentemente usados para tornar os nomes das colunas mais legíveis. Um apelido só existe para a duração da consulta:

```mysql
SELECT column_name
FROM table_name AS alias_name;
```

````python
# Artist LEFT JOIN Album
str_query = sqlalchemy.text("SELECT t1.Name as ArtistName, t2.Title as AlbumTitle FROM Artist as t1 LEFT JOIN Album as t2 where t1.ArtistId = t2.ArtistId")
df = pd.read_sql_query(str_query, con=engine.connect())
print(df)

ex2_9 = df.head().copy()
````

#### Exercício 2.10

Agora filtre a base combinada para recuperar somente os albuns do AC/DC

````python
# Artist LEFT JOIN Album
str_query = sqlalchemy.text("""SELECT t1.Name as ArtistName, t2.Title as AlbumTitle FROM Artist as t1 LEFT JOIN Album as t2 where t1.ArtistId = t2.ArtistId
               and t1.Name like 'AC/DC'""")
df = pd.read_sql_query(str_query, con=engine.connect())
print(df)

ex2_10 = df.head().copy()
````

#### Exercício 2.11

````python
# QUAL TIPO DE MÍDIA TEM MAIS FAIXAS (tracks)?
str_query = sqlalchemy.text("""SELECT t2.Name as media_type, count(t1.Name) as number_tracks
               FROM Track as t1 INNER JOIN MediaType as t2 ON t1.MediaTypeId = t2.MediaTypeId
               GROUP BY media_type
               ORDER BY number_tracks desc""")
df = pd.read_sql_query(str_query, con=engine.connect())
print(df)

ex2_11 = df.head().copy()
````
